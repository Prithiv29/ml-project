```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/train.csv")
dev = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/dev.csv")
test = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/test.csv")
```

```python
plt.figure(figsize=(4,4))
sns.countplot(data=train,x='label')
plt.title("Training distribution")
plt.xlabel('Label')
plt.ylabel('No of comments')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sns.countplot(data=train, x="label", ax=axes[0])
axes[0].set_title("Train")
sns.countplot(data=dev, x="label", ax=axes[1])
axes[1].set_title("Development")
plt.tight_layout()
plt.show() 
```

```python
train["word_count"]=train['text'].str.split().str.len()
plt.figure(figsize=(6,4))
sns.histplot(train["word_count"],bins=500)
plt.title("Word count distribution")
plt.xlabel("No of words")
plt.ylabel("No of comments")
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=train,
    x="label",
    y="word_count"
)
plt.title("Word Count by Class")
plt.xlabel("Label")
plt.ylabel("Number of Words")
plt.show()
```
