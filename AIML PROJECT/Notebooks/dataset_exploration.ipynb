```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python

train = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/train.csv")
dev = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/dev.csv")
test = pd.read_csv("D:/Prithiv/AIML PROJECT/DATASET/test.csv")
```

```python
print(train.shape)
print(dev.shape)
print(test.shape)
```

```python
print("Train:",train.head)
print("Dev:",dev.head)
print("Test:",test.head)
```

```python
print("TRAIN")
train.info()

print("\nDEV")
dev.info()

print("\nTEST")
test.info()
```

```python
print("TRAIN MISSING VALUES")
print(train.isnull().sum())

print("\nDEV MISSING VALUES")
print(dev.isnull().sum())

print("\nTEST MISSING VALUES")
print(test.isnull().sum())
```

```python
print("Duplicated values in train:",train.duplicated().sum())
print("Duplicated values in dev:",dev.duplicated().sum())
print("Duplicated values in test:",test.duplicated().sum())
```

```python
print("Train duplicate IDs:", train["id"].duplicated().sum())
print("Dev duplicate IDs:", dev["id"].duplicated().sum())
print("Test duplicate IDs:", test["id"].duplicated().sum())
```

```python
print(train["label"].value_counts())
print(dev['label'].value_counts())
```

```python
print(train['label'].unique())
print(dev['label'].unique())
```

```python
print("TRAIN")
print(train["label"].value_counts(normalize=True) * 100)

print("\nDEV")
print(dev["label"].value_counts(normalize=True) * 100)
```

```python
print("Empty train texts:",
      train["text"].fillna("").str.strip().eq("").sum())

print("Empty dev texts:",
      dev["text"].fillna("").str.strip().eq("").sum())

print("Empty test texts:",
      test["text"].fillna("").str.strip().eq("").sum())
```

```python
train["word_count"] = train["text"].str.split().str.len()
dev["word_count"] = dev["text"].str.split().str.len()
test["word_count"] = test["text"].str.split().str.len()
print(train['word_count'].describe());
```

```python
print("Train comments with <3 words:",
      (train["word_count"] < 3).sum())

print("Dev comments with <3 words:",
      (dev["word_count"] < 3).sum())

print("Test comments with <3 words:",
      (test["word_count"] < 3).sum())
```
